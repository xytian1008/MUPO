```python
import os
import random

from tqdm import tqdm
from datasets import Dataset, load_dataset, load_from_disk, concatenate_datasets

train_dataset = load_dataset("xytian1008/MUPO-Thinker-train36k_public")['train']
```

```python
mathverse = load_dataset("AI4Math/MathVerse", "testmini")['testmini']

# Only keep the required columns and rename them using map
def process_mathverse(example):
    return {
        'problem': f"<image>{example['question']}",
        'answer': example['answer'],
        'images': [example['image']],
        'source': 'mathverse'
    }

mathverse = mathverse.map(process_mathverse, remove_columns=mathverse.column_names, num_proc=os.cpu_count())
```

```python
mmmu_pro = load_dataset("MMMU/MMMU_Pro", 'standard (4 options)')['test']

def filter_single_image(example):
    # Only keep if image_1 is not empty and all other image columns are empty
    return (
        example['image_1'] and
        all(not example[f'image_{i}'] for i in range(2, 8))
    )

def process_mmmu_pro(example):
    options = eval(example['options'])
    option_labels = ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H']  # extend as needed
    formatted_options = [f"{label}. {opt}" for label, opt in zip(option_labels, options)]
    prompt = example['question'].replace("<image 1>", "<image>") + "\n\n" + "\n".join(formatted_options)
    return {
        'problem': prompt,
        'answer': example['answer'],
        'images': [example['image_1']],
        'source': 'mmmu_pro'
    }

mmmu_pro = mmmu_pro.filter(filter_single_image, num_proc=os.cpu_count())
mmmu_pro = mmmu_pro.map(process_mmmu_pro, remove_columns=mmmu_pro.column_names, num_proc=os.cpu_count())
```

```python
mathvision = load_dataset("MathLLMs/MathVision")['testmini']

def process_mathvision(example):
    question = example['question'].replace("<image1>", "")
    question = f"<image>{question}"
    return {
        'problem': question,
        'answer': example['answer'],
        'images': [example['decoded_image']],
        'source': 'mathvision'
    }

mathvision = mathvision.map(process_mathvision, remove_columns=mathvision.column_names, num_proc=os.cpu_count())
```

```python
import math

# List of all datasets
datasets = [mathverse, logicvista, geo3k, wemath, mathvista, mmmu_pro, mathvision, mmstar, hallbench, mmvet]

# Compute the total number of examples across all datasets
total_examples = sum(len(d) for d in datasets if len(d) > 0)
target_n = int(0.55 * total_examples)

# Compute number of examples to select from each dataset to distribute as equally as possible
# For small datasets, take all examples; for others, distribute remaining quota

# First, store lengths
lengths = [len(d) for d in datasets]
remaining = target_n
selected_examples = []
samples_per_dataset = []

# First pass: take all from really small datasets if possible, mark those as fixed.
fixed_indices = []
for idx, l in enumerate(lengths):
    # Heuristic: If a dataset < avg quota, take all. Will rebalance exact quotas below.
    avg_quota = target_n // len(datasets)
    if l <= avg_quota:
        samples_per_dataset.append(l)
        remaining -= l
        fixed_indices.append(idx)
    else:
        samples_per_dataset.append(None)  # To be filled in below

# Number of datasets left to divide into
num_remaining_datasets = samples_per_dataset.count(None)

# Second pass: divide remaining quota equally among remaining
for idx, val in enumerate(samples_per_dataset):
    if val is None:
        l = lengths[idx]
        equal_share = remaining // num_remaining_datasets
        n_this = min(l, equal_share)
        samples_per_dataset[idx] = n_this
        remaining -= n_this
        num_remaining_datasets -= 1

# If there's any 'remaining' quota (e.g. by rounding-down), randomly (+1) to some datasets that aren't at max
if remaining > 0:
    for idx, n_this in enumerate(samples_per_dataset):
        if n_this < lengths[idx]:
            additional_take = min(remaining, lengths[idx] - n_this)
            samples_per_dataset[idx] += additional_take
            remaining -= additional_take
            if remaining <= 0:
                break

# Now, select the respective number of examples from each dataset
import random

selected_examples = []
for dataset, n_samples in zip(datasets, samples_per_dataset):
    n_avail = len(dataset)
    if n_samples == 0:
        continue
    if n_samples >= n_avail:
        indices = list(range(n_avail))
    else:
        indices = random.sample(range(n_avail), n_samples)
    selected_examples.append(dataset.select(indices))

all_val_datasets = concatenate_datasets(selected_examples)
all_val_datasets = all_val_datasets.shuffle(seed=0)
```

```python
# Remove examples where "<image>" appears more than once in the 'problem' field
def has_single_image_token(example):
    return example['problem'].count("<image>") == 1

all_val_datasets = all_val_datasets.filter(has_single_image_token, num_proc=os.cpu_count())
```

```python
# Remove 'source' column from all_val_datasets
if 'source' in all_val_datasets.column_names:
    all_val_datasets = all_val_datasets.remove_columns('source')

# Remove random examples from train_dataset and add all_val_datasets to it
import numpy as np

num_replace = len(all_val_datasets)
if num_replace > len(train_dataset):
    raise ValueError("all_val_datasets is larger than train_dataset; cannot remove more examples than exist in train_dataset.")

# Randomly select indices to delete from train_dataset
remove_indices = np.random.choice(len(train_dataset), size=num_replace, replace=False)
keep_indices = [i for i in range(len(train_dataset)) if i not in set(remove_indices)]

# Keep only the non-removed examples
train_dataset = train_dataset.select(keep_indices)

train_dataset = concatenate_datasets([train_dataset, all_val_datasets])
train_dataset = train_dataset.shuffle(seed=0)
```
